**Задание 14 (0.5 балла).** Используя функцию ```gen_uncertain_data``` для генерации выборки, отобразите на графике синим цветом функцию ```y(x)```, а также ее доверительный интервал в виде закрашенной зеленым цветом области от ```y[i] - error[i]``` до ```y[i] + error[i]```. Полезной может оказаться функция [fill_between](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.fill_between.html).  

````python
def gen_uncertain_data(seed=None):
    rgn = np.random.default_rng(seed=seed)
    x = np.linspace(0, 30, 100)
    y = np.sin(x/6*np.pi) + rgn.normal(0, 0.02, size=x.shape)
    error = rgn.normal(0.1, 0.02, size=y.shape)
    return x, y, error
````
