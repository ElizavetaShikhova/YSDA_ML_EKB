На семинаре мы кратко познакомились с такой моделью, как SVM. Метод опорных векторов -- модель линейной классификации, в основе которой используется функция ошибки вида:

$$
L(y, a_{w, w_0}(x)) = (1-y(wx+w_0))_+ \frac{1}{2C} \Vert w \Vert= \max(0, 1-y(wx+b)) + \frac{1}{2C} \Vert w \Vert \to \min_{w, w_0}
$$

Неплохо, не правда ли? В терминологии Воронцова подобное преобразование $\psi: X \to H$ называется **спрямляющим**. А что если спрямляющее преобразование $\psi$ представляет из себя сложновычислимую функцию или вообще неизвестно? Kernel trick comes to action. 

В **двойственной задаче** метода опорных векторов, можно построить разделяющую поверхность не в терминах разделяющей гиперплоскости $w$ в спрямляемом пространстве, а в терминах ядерной функции $K$ и некоторых коэффициентов $\lambda$. В такой постановке оказывается необязательно знать $w$ и спрямляющее преобразование, а достаточно знать такую функцию $K: X \to \mathbb{R}$, которая будет являться скалярным произведением в пространстве $H$, т.е. $K(x, x')=\psi(x) \cdot \psi(x')$.
